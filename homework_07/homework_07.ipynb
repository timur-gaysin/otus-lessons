### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
#cсоздание массива
arr = np.arange(10,50)
#разворачиваем
reversed_arr = arr[::-1]
#поиск четных чисел
even_numbers = reversed_arr[reversed_arr % 2 ==0]
#проверка
print("Исходный массив")
print(arr)
print("Развернутый массив")
print(reversed_arr)
print("Четные элементы")
print(even_numbers)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
#генерация массива
arr = np.random.randint(0,100, size = (4,3,2))

print("Массив:")
print(arr)

print("Минимум:", arr.min())
print("Максимум:" , arr.max())
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
arr1 = np.random.randint(0,10,size= (6,4))
arr2 = np.random.randint(0,10,size = (4,3))

multArray = np.matmul(arr1,arr2)

print("Матрица 1:")
print(arr1)

print("Матрица 2:")
print(arr2)

print("Результат умножения")
print(multArray)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
#генерация массива

arr = np.random.randint(0,100, size = (7,7))

print("Исходный массив")
print(arr)

#среднее значение
mean = arr.mean()

#стандартное отклонение
std = arr.std()

print("Среднее: ",mean)
print("Стандартное отклонение: ", std)

#нормализация
normalized = (arr-mean) / std

print("Нормализованный массив:")
print(normalized)
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
tips = sns.load_dataset("tips")
average_bill = tips.groupby("day")["total_bill"].mean()
print(average_bill)
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
tips = sns.load_dataset("tips")
median_bill = tips["total_bill"].median()
result = (
    tips[tips["total_bill"] > median_bill]
    .groupby("sex")["tip"]
    .mean()
)
print(result)
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(data=tips, x="total_bill")

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x = "total_bill",
    y = "tip"
)
plt.title("Total Bill & Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

### Постройте pairplot

```python
tips = sns.load_dataset("tips")

sns.pairplot(tips)

plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x = "day",
    y = "total_bill"
)

plt.title("Total Bill by Day")
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
tips = sns.load_dataset("tips")

g = sns.FacetGrid(tips, col = "time")

g.map(
    sns.histplot,
    "tip"
)

plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
tips = sns.load_dataset("tips")

g = sns.FacetGrid(
    tips,
    col = "sex",
    hue = "smoker",
    height=5
)

g.map(
    sns.scatterplot,
    "total_bill",
    "tip"
)

g.add_legend()

plt.show()
```
