```python
#ищем числовые признаки
features = df.select_dtypes(include="number")
#вычисляем матрицу корреляции
corr_matrix = features.corr(method="pearson")
corr_matrix
```

```python
#4. строим heatmap для матрицы корреляций

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

mask = np.triu(
    np.ones_like(corr_matrix, dtype=bool)
)

plt.figure(figsize=(15, 12))

sns.heatmap(
    corr_matrix,
    mask=mask,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    annot=False
)

plt.title("Корреляции между числовыми признаками")
plt.tight_layout()
plt.show()
```

```python
#для поиска связей выберем порог равный 0.9
threshold = 0.9

pairs = []

columns = features.columns

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        r = corr_matrix.iloc[i, j]

        if abs(r) > threshold:
            pairs.append({
                "feature_1": columns[i],
                "feature_2": columns[j],
                "correlation": r
            })

strong_pairs = pd.DataFrame(
    pairs,
    columns=["feature_1", "feature_2", "correlation"]
)

strong_pairs = strong_pairs.sort_values(
    by="correlation",
    key=lambda column: column.abs(),
    ascending=False
)

strong_pairs
```

```python
'''
Как моджно увидеть, признаки с высокой корреляцией относятся к числовым признакам: площади, периметру и радиусу
что неудивительно, така как в математике площадь, периметр и радиус тесно связаны между собой.
'''
```

```python
#Часть 2. Моделирование при помощи kNN
#1. Разбейте данные на train-test, отложив 30% выборки для тестирования.
#x - таблица из 30 признаков, y - правильные ответы

X = df.drop(columns=["diagnosis"])
y = df["diagnosis"]

#разбивка данных - 30% на тестовую выборку, 70% на обучающую выборку
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("Доли классов в train:")
print(y_train.value_counts(normalize=True))

print("\nДоли классов в test:")
print(y_test.value_counts(normalize=True))
```

```python
#2.Приведите все непрерывные переменные к одному масштабу при помощи стандартизации. Кратко поясните, почему стандартизация здесь нужна.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(X_train_scaled.mean(axis=0).round(3))
print(X_train_scaled.std(axis=0).round(3))

'''
Без выполнения стандартизации признаки с большим масштабом будут доминировать над признаками с меньшим масштабом, что может привести к неправильной работе алгоритма kNN. Стандартизация позволяет привести все признаки к одному масштабу, что улучшает качество модели.
'''
```

```python
#3. Постройте модель kNN «из коробки» без настройки параметров.
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier()

knn.fit(X_train_scaled, y_train)
y_pred = knn.predict(X_test_scaled)
comparison = pd.DataFrame({
    "Настоящий диагноз": y_test.to_numpy(),
    "Предсказание": y_pred
})

comparison.head(10)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["B", "M"]
)

display(pd.DataFrame(
    cm,
    index=["На самом деле B", "На самом деле M"],
    columns=["Предсказан B", "Предсказан M"]
))
```

```python
#метрики

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

metrics = pd.Series({
    "accuracy": accuracy_score(y_test, y_pred),
    "precision": precision_score(y_test, y_pred, pos_label="M"),
    "recall": recall_score(y_test, y_pred, pos_label="M"),
    "f1_score": f1_score(y_test, y_pred, pos_label="M")
})

display(metrics.round(4))

probabilities = knn.predict_proba(X_test_scaled)

print("Порядок классов:", knn.classes_)
print(probabilities[:5])

```

```python
#ROC
m_index = list(knn.classes_).index("M")

y_score = probabilities[:, m_index]
y_test_binary = (y_test == "M").astype(int)

from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(
    y_test_binary,
    y_score
)

roc_auc = roc_auc_score(
    y_test_binary,
    y_score
)

print(f"ROC AUC: {roc_auc:.4f}")

plt.figure(figsize=(6, 5))

plt.plot(
    fpr,
    tpr,
    marker="o",
    label=f"kNN, AUC = {roc_auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="gray",
    label="Уровень случайного ранжирования"
)

plt.xlabel("FPR — доля B, ошибочно названных M")
plt.ylabel("TPR — доля обнаруженных M")
plt.title("ROC-кривая kNN на тестовой выборке")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
#Выводы по пункту 2-3
'''
Все 107 наблюдений B распознаны правильно.Из 64 наблюдений M обнаружены 58, пропущены 6. Модель ни разу не назвала B злокачественным.
Что показывает ROC кривая - проходит близко к верхнему левому углу, что говорит о высокой точности модели. AUC = 0.988, что также подтверждает высокую точность модели.
'''
```

```python
#4 проведем настройку параметра числа соседей на кросс-валидации и оценим улучшились ли результаты.
# Мы брали количество соседей = 5, попробуем перебором посмотреть результаты при других значениях числа соседей.

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier())
])

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import make_scorer, f1_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

f1_m = make_scorer(
    f1_score,
    pos_label="M"
)

param_grid = {
    "knn__n_neighbors": list(range(1, 31))
}

search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring=f1_m,
    cv=cv,
    refit=True
)

search.fit(X_train, y_train)
```

```python
#посмотрим на результат
print("Лучшие параметры:", search.best_params_)
print("Средняя F1 на кросс-валидации:", search.best_score_)
```

```python
#Зависимость качества от К
cv_results = pd.DataFrame(search.cv_results_)

cv_table = pd.DataFrame({
    "k": cv_results["param_knn__n_neighbors"].astype(int),
    "mean_f1": cv_results["mean_test_score"],
    "std_f1": cv_results["std_test_score"]
}).sort_values("k")

display(cv_table)
```

```python
#построение графика
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.errorbar(
    cv_table["k"],
    cv_table["mean_f1"],
    yerr=cv_table["std_f1"],
    fmt="o-",
    capsize=3
)

plt.xlabel("Число соседей k")
plt.ylabel("Средняя F1 для класса M")
plt.title("Подбор k на кросс-валидации")
plt.grid(alpha=0.3)
plt.show()
```

```python
#Проверка выбранной модели на тесте 
best_model = search.best_estimator_

y_pred_tuned = best_model.predict(X_test)

m_index = list(best_model.classes_).index("M")
y_score_tuned = best_model.predict_proba(X_test)[:, m_index]

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Ответы исходной модели с k = 5
y_pred_base = knn.predict(X_test_scaled)

base_m_index = list(knn.classes_).index("M")
y_score_base = knn.predict_proba(X_test_scaled)[:, base_m_index]

y_test_binary = (y_test == "M").astype(int)


def evaluate(y_true, y_pred, y_score):
    return {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, pos_label="M"),
        "recall": recall_score(y_true, y_pred, pos_label="M"),
        "f1_score": f1_score(y_true, y_pred, pos_label="M"),
        "roc_auc": roc_auc_score(
            (y_true == "M").astype(int),
            y_score
        )
    }


comparison = pd.DataFrame({
    "До настройки": evaluate(y_test, y_pred_base, y_score_base),
    "После настройки": evaluate(y_test, y_pred_tuned, y_score_tuned)
})

comparison["Изменение"] = (
    comparison["После настройки"] - comparison["До настройки"]
)

display(comparison.round(4))
```

```python
#Бонусное задание
#1. Уберите переменные, корреляция Пирсона между которыми превышает 0.85.
corr_matrix = X_train.corr(method="pearson").abs()
threshold = 0.85

kept_features = []
drop_details = []

for feature in corr_matrix.columns:
    for kept_feature in kept_features:
        correlation = corr_matrix.loc[feature, kept_feature]

        if correlation > threshold:
            drop_details.append({
                "Удалённый признак": feature,
                "Сохранённый признак": kept_feature,
                "|r|": correlation
            })
            break
    else:
        kept_features.append(feature)
```

```python
#Оценим результат
drop_table = pd.DataFrame(
    drop_details,
    columns=["Удалённый признак", "Сохранённый признак", "|r|"]
)

display(drop_table.round(3))

print("Было признаков:", X_train.shape[1])
print("Осталось признаков:", len(kept_features))
```

```python
X_train_selected = X_train[kept_features].copy()
X_test_selected = X_test[kept_features].copy()

y_train_binary = (y_train == "M").astype(int)
y_test_binary = (y_test == "M").astype(int)

selected_corr = X_train_selected.corr().abs()

upper_mask = np.triu(
    np.ones(selected_corr.shape, dtype=bool),
    k=1
)

remaining_pairs = selected_corr.where(upper_mask).stack()

print(
    "Количество оставшихся пар с |r| > 0.85:",
    (remaining_pairs > threshold).sum()
)
```

```python
#2 Постройте модель логистической регрессии «из коробки» без настройки параметров. Оцените метрики качества бинарной классификации (accuracy, precision, recall, f1_score), постройте ROC-кривую и посчитайте площадь под ней. Визуализируйте влияние коэффициентов на целевую переменную.

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

logreg_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression())
])

logreg_pipeline.fit(X_train_selected, y_train)
```

```python
#подсчет метрик
y_pred_logreg = logreg_pipeline.predict(X_test_selected)

m_index = list(logreg_pipeline.classes_).index("M")

y_score_logreg = logreg_pipeline.predict_proba(
    X_test_selected
)[:, m_index]

import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

y_test_binary = (y_test == "M").astype(int)

logreg_metrics = pd.Series({
    "accuracy": accuracy_score(y_test, y_pred_logreg),
    "precision": precision_score(
        y_test, y_pred_logreg, pos_label="M"
    ),
    "recall": recall_score(
        y_test, y_pred_logreg, pos_label="M"
    ),
    "f1_score": f1_score(
        y_test, y_pred_logreg, pos_label="M"
    ),
    "roc_auc": roc_auc_score(
        y_test_binary, y_score_logreg
    )
})

display(logreg_metrics.round(4))
```

```python
#изучим ошибки

cm_logreg = confusion_matrix(
    y_test,
    y_pred_logreg,
    labels=["B", "M"]
)

display(pd.DataFrame(
    cm_logreg,
    index=["На самом деле B", "На самом деле M"],
    columns=["Предсказан B", "Предсказан M"]
))
```

```python
#3. проведите настройку параметра регуляризации C на кросс-валидации (LogisticRegressionCV). Снова оцените результаты и постройте визуализацию влияния коэффициентов на целевую переменную.

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, StratifiedKFold


#стандартизация признаков


cv_lr = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    (
        "model",
        LogisticRegression(
            solver="lbfgs",
            max_iter=2000
        )
    )
])


#настройка С
c_values = np.logspace(-4, 4, 17)


params = {
    "model__C": c_values
}

logreg_cv = GridSearchCV(
    estimator=pipeline,
    param_grid=params,
    cv=cv_lr,
    scoring="f1",
    refit=True
)

logreg_cv.fit(
    X_train_selected,
    y_train_binary
)
```

```python
# Смотрим результаты подбора

import pandas as pd

best_c = logreg_cv.best_params_["model__C"]

cv_results = pd.DataFrame(
    logreg_cv.cv_results_
)

c_table = cv_results[
    [
        "param_model__C",
        "mean_test_score",
        "std_test_score"
    ]
].copy()

c_table.columns = [
    "C",
    "Средняя F1",
    "Стандартное отклонение F1"
]

print("Выбранное C:", best_c)

display(
    c_table.round(5)
)
```

```python
#Подбор С на кросс-валидации

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.errorbar(
    c_table["C"],
    c_table["Средняя F1"],
    yerr=c_table["Стандартное отклонение F1"],
    fmt="o-",
    capsize=3
)

plt.xscale("log")
plt.axvline(best_c, color="orange", linestyle="--",
            label=f"Выбранное C = {best_c:.4g}")

plt.xlabel("C — обратная сила регуляризации")
plt.ylabel("Средняя F1 для M")
plt.title("Подбор C на кросс-валидации")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# оценка настроенной модели на тесте

y_pred_lr_cv = logreg_cv.predict(
    X_test_selected
)

# классы лежат внутри LogisticRegression,
# которая находится внутри Pipeline
classes = (
    logreg_cv
    .best_estimator_
    .named_steps["model"]
    .classes_
)

m_index = list(classes).index(1)

y_score_lr_cv = logreg_cv.predict_proba(
    X_test_selected
)[:, m_index]

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

def evaluate_binary(y_true, y_pred, y_score):
    return {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall_score(y_true, y_pred, zero_division=0),
        "f1_score": f1_score(y_true, y_pred, zero_division=0),
        "roc_auc": roc_auc_score(y_true, y_score)
    }

metrics_lr_cv = evaluate_binary(
    y_test_binary,
    y_pred_lr_cv,
    y_score_lr_cv
)

display(pd.Series(metrics_lr_cv).round(4))
```

```python
#матрица ошибок 
cm_lr_cv = confusion_matrix(
    y_test_binary,
    y_pred_lr_cv,
    labels=[0, 1]
)

display(pd.DataFrame(
    cm_lr_cv,
    index=["На самом деле B", "На самом деле M"],
    columns=["Предсказан B", "Предсказан M"]
))
```

```python
#Сравнение моделей в одной таблице

final_comparison = pd.DataFrame({
    "kNN после настройки": evaluate_binary(
        y_test_binary,
        (y_pred_tuned == "M").astype(int),
        y_score_tuned
    ),
    "Логрегрессия до настройки": evaluate_binary(
        y_test_binary,
        (y_pred_logreg == "M").astype(int),
        y_score_logreg
    ),
    "Логрегрессия после настройки": metrics_lr_cv
})

display(final_comparison.round(4))
```

```python
#Сравнение ROC кривых

from sklearn.metrics import RocCurveDisplay

fig, ax = plt.subplots(figsize=(7, 6))

RocCurveDisplay.from_predictions(
    y_test_binary,
    y_score_tuned,
    name="Настроенный kNN",
    ax=ax
)

RocCurveDisplay.from_predictions(
    y_test_binary,
    y_score_lr_cv,
    name="Настроенная логистическая регрессия",
    ax=ax
)

ax.plot([0, 1], [0, 1], "--", color="gray")
ax.set_xlabel("FPR — доля B, ошибочно названных M")
ax.set_ylabel("TPR — доля обнаруженных M")
ax.set_title("ROC-кривые настроенных моделей")
ax.grid(alpha=0.3)
plt.show()
```
