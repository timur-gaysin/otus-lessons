### Задание 4. Минимизация суммы отклонений

Даны три точки:

$$
(-3, 2), \quad (2, 2), \quad (5, -3)
$$

Необходимо найти прямую, проходящую через точку $(0,0)$, для которой сумма отклонений данных точек по оси $y$ минимальна.

#### Шаг 1. Зададим параметры прямой

Так как прямая обязательно проходит через начало координат $(0,0)$, её можно задать одним параметром $k$:

$$
y = kx
$$

где $k$ — коэффициент наклона прямой.

Для каждой из трёх точек найдём отклонение фактического значения $y$ от значения, предсказанного прямой $kx$.

Для точки $(-3,2)$:

$$
e_1 = 2 - k(-3) = 2 + 3k
$$

Для точки $(2,2)$:

$$
e_2 = 2 - 2k
$$

Для точки $(5,-3)$:

$$
e_3 = -3 - 5k
$$

#### Шаг 2. Запишем целевую функцию

Необходимо минимизировать сумму абсолютных отклонений:

$$
F(k) = |2+3k| + |2-2k| + |-3-5k|
$$

Таким образом, необходимо найти:

$$
\min_k F(k)
$$

#### Шаг 3. Найдём оптимальный параметр

Найдём значения $k$, при которых отдельные отклонения обращаются в ноль.

$$
2+3k=0
\quad\Rightarrow\quad
k=-\frac{2}{3}
$$

$$
2-2k=0
\quad\Rightarrow\quad
k=1
$$

$$
-3-5k=0
\quad\Rightarrow\quad
k=-\frac{3}{5}
$$

Для суммы абсолютных отклонений минимум достигается при:

$$
k=-\frac{3}{5}
$$

Следовательно, искомая прямая:

$$
\boxed{y=-\frac{3}{5}x}
$$

Проверим значение целевой функции:

$$
F\left(-\frac35\right)
=
\left|2+3\left(-\frac35\right)\right|
+
\left|2-2\left(-\frac35\right)\right|
+
\left|-3-5\left(-\frac35\right)\right|
$$

$$
=
\left|\frac15\right|
+
\left|\frac{16}{5}\right|
+
|0|
=
\frac{17}{5}
$$

$$
\boxed{F_{\min}=\frac{17}{5}=3.4}
$$

#### Шаг 4. Результат

Оптимальная прямая для минимальной суммы абсолютных отклонений:

$$
\boxed{y=-\frac35x}
$$

Минимальная сумма отклонений:

$$
\boxed{3.4}
$$

### Задание 5. Минимизация суммы квадратов отклонений

Используем те же точки:

$$
(-3,2), \quad (2,2), \quad (5,-3)
$$

Необходимо найти прямую, проходящую через $(0,0)$, для которой сумма квадратов отклонений данных точек по оси $y$ минимальна.

#### Шаг 1. Зададим параметры прямой

Прямая проходит через начало координат, поэтому:

$$
y=kx
$$

где $k$ — единственный параметр прямой.

Отклонения для наших точек:

$$
e_1=2+3k
$$

$$
e_2=2-2k
$$

$$
e_3=-3-5k
$$

#### Шаг 2. Запишем целевую функцию

Теперь необходимо минимизировать не абсолютные значения отклонений, а их квадраты:

$$
F(k)=(2+3k)^2+(2-2k)^2+(-3-5k)^2
$$

Раскроем скобки:

$$
(2+3k)^2=4+12k+9k^2
$$

$$
(2-2k)^2=4-8k+4k^2
$$

$$
(-3-5k)^2=9+30k+25k^2
$$

Сложим полученные выражения:

$$
F(k)
=
4+12k+9k^2
+
4-8k+4k^2
+
9+30k+25k^2
$$

Получаем:

$$
F(k)=38k^2+34k+17
$$

#### Шаг 3. Найдём минимум целевой функции

Для нахождения минимума найдём производную:

$$
F'(k)=76k+34
$$

Приравняем её к нулю:

$$
76k+34=0
$$

$$
76k=-34
$$

$$
k=-\frac{34}{76}=-\frac{17}{38}
$$

Приблизительное значение:

$$
k\approx-0.447
$$

Так как коэффициент при $k^2$ положительный:

$$
38>0
$$

функция является параболой, направленной вверх, поэтому найденная точка действительно является минимумом.

Следовательно, искомая прямая:

$$
\boxed{y=-\frac{17}{38}x}
$$

#### Шаг 4. Результат

Оптимальная прямая по методу наименьших квадратов:

$$
\boxed{y=-\frac{17}{38}x}
$$

или приблизительно:

$$
\boxed{y\approx-0.447x}
$$

Таким образом, минимизация суммы абсолютных отклонений и минимизация суммы квадратов отклонений дают разные оптимальные прямые:

$$
\text{Задание 4:}\quad y=-\frac35x
$$

$$
\text{Задание 5:}\quad y=-\frac{17}{38}x
$$

### Задание 7.

На рисунке изображена плотность некоторого непрерывного распределения:

<img src="https://github.com/dmi3eva/resources/blob/main/2026-04-27_PDF.png?raw=true" width="250">

**7-1:** Вычислите вероятность попадания случайной величины в интервал от 0 до 0.5 включительно.

**7-2:** Запишите функцию вероятности (CDF) этого распределения.








### Решение

Дана плотность непрерывного распределения:

$$
f(x)=2x,\qquad 0\le x\le1
$$

За пределами интервала $[0,1]$ плотность равна нулю.

Проверим, что функция является плотностью распределения. Площадь под графиком должна быть равна 1:

$$
\int_0^1 2x\,dx
=
\left.x^2\right|_0^1
=
1
$$

#### 7-1. Вероятность попадания в интервал от 0 до 0.5

Для непрерывного распределения вероятность попадания в интервал
равна площади под графиком плотности на этом интервале:

$$
P(0\le X\le0.5)
=
\int_0^{0.5}2x\,dx
$$

Вычислим интеграл:

$$
\int 2x\,dx=x^2
$$

Следовательно:

$$
P(0\le X\le0.5)
=
\left.x^2\right|_0^{0.5}
$$

$$
P(0\le X\le0.5)
=
0.5^2-0^2
=
0.25
$$

Таким образом:

$$
\boxed{P(0\le X\le0.5)=0.25}
$$

или 25%.

---

#### 7-2. Функция распределения (CDF)

Функция распределения определяется как:

$$
F(x)=P(X\le x)
$$

или через плотность:

$$
F(x)=\int_{-\infty}^{x}f(t)\,dt
$$

Рассмотрим три случая.

Если $x\le0$, то:

$$
F(x)=0
$$

Если $0<x<1$, то:

$$
F(x)=\int_0^x2t\,dt
$$

$$
F(x)=\left.t^2\right|_0^x=x^2
$$

Если $x\ge1$, вся возможная область распределения уже учтена, поэтому:

$$
F(x)=1
$$

Таким образом, функция распределения имеет вид:

$$
\boxed{
F(x)=
\begin{cases}
0, & x\le0,\\
x^2, & 0<x<1,\\
1, & x\ge1.
\end{cases}
}
$$

### Вернемся к исходным данным задачи 5

Даны три точки:

$$
(-3,2),\qquad (2,2),\qquad (5,-3)
$$

Необходимо подобрать прямую из множества всех прямых на плоскости методом наименьших квадратов, используя формулу:

$$
w_{min}=(X^TX)^{-1}X^Ty
$$

#### Шаг 1. Зададим уравнение прямой

Так как рассматриваются все прямые на плоскости, прямая имеет вид:

$$
y=w_0+w_1x
$$

где:

- $w_0$ — свободный член;
- $w_1$ — коэффициент наклона прямой.

Вектор параметров:

$$
w=
\begin{pmatrix}
w_0\\
w_1
\end{pmatrix}
$$

Для данных точек составим матрицу признаков $X$.

Первый столбец состоит из единиц и соответствует свободному члену $w_0$:

$$
X=
\begin{pmatrix}
1 & -3\\
1 & 2\\
1 & 5
\end{pmatrix}
$$

Вектор наблюдаемых значений $y$:

$$
y=
\begin{pmatrix}
2\\
2\\
-3
\end{pmatrix}
$$

---

#### Шаг 2. Найдём $X^T$

$$
X^T=
\begin{pmatrix}
1 & 1 & 1\\
-3 & 2 & 5
\end{pmatrix}
$$

---

#### Шаг 3. Вычислим $X^TX$

$$
X^TX=
\begin{pmatrix}
1 & 1 & 1\\
-3 & 2 & 5
\end{pmatrix}
\begin{pmatrix}
1 & -3\\
1 & 2\\
1 & 5
\end{pmatrix}
$$

Получаем:

$$
X^TX=
\begin{pmatrix}
3 & 4\\
4 & 38
\end{pmatrix}
$$

---

#### Шаг 4. Найдём обратную матрицу $(X^TX)^{-1}$

Для матрицы

$$
\begin{pmatrix}
a & b\\
c & d
\end{pmatrix}
$$

обратная матрица вычисляется по формуле:

$$
\begin{pmatrix}
a & b\\
c & d
\end{pmatrix}^{-1}
=
\frac{1}{ad-bc}
\begin{pmatrix}
d & -b\\
-c & a
\end{pmatrix}
$$

Определитель матрицы $X^TX$:

$$
\det(X^TX)=3\cdot38-4\cdot4
$$

$$
\det(X^TX)=114-16=98
$$

Следовательно:

$$
(X^TX)^{-1}
=
\frac{1}{98}
\begin{pmatrix}
38 & -4\\
-4 & 3
\end{pmatrix}
$$

---

#### Шаг 5. Вычислим $X^Ty$

$$
X^Ty=
\begin{pmatrix}
1 & 1 & 1\\
-3 & 2 & 5
\end{pmatrix}
\begin{pmatrix}
2\\
2\\
-3
\end{pmatrix}
$$

Первый элемент:

$$
2+2-3=1
$$

Второй элемент:

$$
(-3)\cdot2+2\cdot2+5\cdot(-3)
$$

$$
=-6+4-15=-17
$$

Таким образом:

$$
X^Ty=
\begin{pmatrix}
1\\
-17
\end{pmatrix}
$$

---

#### Шаг 6. Найдём параметры прямой

Используем формулу МНК:

$$
w_{min}=(X^TX)^{-1}X^Ty
$$

Подставляем найденные значения:

$$
w_{min}
=
\frac{1}{98}
\begin{pmatrix}
38 & -4\\
-4 & 3
\end{pmatrix}
\begin{pmatrix}
1\\
-17
\end{pmatrix}
$$

Выполняем умножение:

$$
w_{min}
=
\frac{1}{98}
\begin{pmatrix}
38+68\\
-4-51
\end{pmatrix}
$$

$$
w_{min}
=
\frac{1}{98}
\begin{pmatrix}
106\\
-55
\end{pmatrix}
$$

Следовательно:

$$
w_0=\frac{106}{98}=\frac{53}{49}
$$

$$
w_1=-\frac{55}{98}
$$

В десятичном виде:

$$
w_0\approx1.082
$$

$$
w_1\approx-0.561
$$

---

#### Шаг 7. Запишем найденную прямую

Так как

$$
y=w_0+w_1x
$$

получаем:

$$
\boxed{
y=\frac{53}{49}-\frac{55}{98}x
}
$$

или приближённо:

$$
\boxed{
y\approx1.082-0.561x
}
$$

Таким образом, методом наименьших квадратов из множества всех прямых была найдена прямая:

$$
\boxed{
y=\frac{53}{49}-\frac{55}{98}x
}
$$
